#The Titanic dataset contains information about passengers who travelled on the Titanic ship. It includes details such as passenger class, gender, age, ticket fare, family members and survival status.

The dataset contains both numerical and categorical features. It is useful for understanding data preparation, identifying missing values, cleaning inconsistent records and performing exploratory data analysis

```python
# Separating numerical and categorical columns

numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

categorical_columns = df.select_dtypes(include=['object', 'category', 'bool']).columns

print("Numerical Columns:")
print(list(numerical_columns))

print("\nCategorical Columns:")
print(list(categorical_columns))
```

```python
# Summary statistics of numerical columns

df.describe()
```

```python
# Summary statistics of categorical columns

df.describe(include='all')
```

```python
# Checking missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percentage
})

print("Missing Value Report:")

display(missing_report)
```

```python
# Visualizing missing values

plt.figure(figsize=(12, 6))

missing_values[missing_values > 0].plot(
    kind='bar',
    color='skyblue'
)

plt.title("Missing Values in Titanic Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Creating a copy of the original dataset

cleaned_df = df.copy()

print("Original dataset copied successfully!")

print("Original Shape:", df.shape)
print("Copied Dataset Shape:", cleaned_df.shape)
```

```python
# Handling missing values in numerical columns

age_median = cleaned_df['age'].median()

cleaned_df['age'] = cleaned_df['age'].fillna(age_median)

print("Missing Age Values Handled Successfully!")

print("Remaining Missing Age Values:",
      cleaned_df['age'].isnull().sum())
```

```python
cleaned_df = df.copy()
# Handling missing values in categorical columns

categorical_missing_columns = ['embarked', 'embark_town']

for column in categorical_missing_columns:

    mode_value = cleaned_df[column].mode()[0]

    cleaned_df[column] = cleaned_df[column].fillna(mode_value)

print("Categorical Missing Values Handled Successfully!")
```

```python
# Removing deck column due to excessive missing values

cleaned_df = cleaned_df.drop(columns=['deck'])

print("Deck column removed successfully!")

print("Updated Dataset Shape:", cleaned_df.shape)
```

```python
# Checking missing values after handling

print("Missing Values After Cleaning:")

print(cleaned_df.isnull().sum())
```

```python
# Checking duplicate records

duplicate_count = cleaned_df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)
```

```python
# Removing duplicate records

cleaned_df = cleaned_df.drop_duplicates()

print("Duplicate Records Removed Successfully!")

print("Updated Dataset Shape:", cleaned_df.shape)
```

```python
# Checking unique values in categorical columns

for column in cleaned_df.select_dtypes(include=['object']).columns:

    print("\nColumn:", column)

    print(cleaned_df[column].unique())
```

```python
# Cleaning categorical text values

for column in cleaned_df.select_dtypes(include=['object']).columns:

    cleaned_df[column] = cleaned_df[column].str.strip()

    cleaned_df[column] = cleaned_df[column].str.lower()

print("Categorical Values Standardized Successfully!")
```

```python
# Creating a separate dataset for encoding

encoded_df = cleaned_df.copy()

# Removing redundant columns

columns_to_remove = [
    'class',
    'embark_town',
    'who',
    'adult_male',
    'alive'
]

encoded_df = encoded_df.drop(columns=columns_to_remove)

print("Redundant Columns Removed Successfully!")

print(encoded_df.head())
```

```python
# Identifying categorical columns

categorical_features = encoded_df.select_dtypes(
    include=['object', 'category', 'bool']
).columns.tolist()

print("Categorical Features:")
print(categorical_features)
```

```python
# Applying One-Hot Encoding

encoded_df = pd.get_dummies(
    encoded_df,
    columns=categorical_features,
    dtype=int
)

print("One-Hot Encoding Completed Successfully!")

display(encoded_df.head())
```

```python
# Display final column names after encoding

print("Columns After Encoding:")

print(encoded_df.columns.tolist())
```

```python
# Creating a copy for scaling

scaled_df = encoded_df.copy()

# Selecting numerical feature columns

numerical_features = [
    'pclass',
    'age',
    'sibsp',
    'parch',
    'fare'
]

print("Numerical Features Selected:")
print(numerical_features)
```

```python
# Applying StandardScaler

scaler = StandardScaler()

scaled_df[numerical_features] = scaler.fit_transform(
    scaled_df[numerical_features]
)

print("Standard Scaling Completed Successfully!")
```

```python
# Display scaled data

scaled_df.head()
```

```python
# Checking mean and standard deviation after scaling

print("Mean of Scaled Numerical Features:")

print(scaled_df[numerical_features].mean())

print("\nStandard Deviation of Scaled Numerical Features:")

print(scaled_df[numerical_features].std())
```

StandardScaler transforms numerical features so that they have approximately zero mean and unit standard deviation.

Scaling is useful for machine learning algorithms that are sensitive to feature magnitudes, such as K-Nearest Neighbors, K-Means and Logistic Regression.

Identifier columns and the survived target variable are not scaled.

Since this activity focuses on dataset preparation rather than model training, scaling is demonstrated on the prepared dataset. In an actual predictive workflow, scaling parameters should be learned from training data only.

```python
# Checking final dataset shape

print("Final Dataset Shape:")
print(scaled_df.shape)
```

```python
# Checking remaining missing values

print("Remaining Missing Values:")

print(scaled_df.isnull().sum().sum())
```

```python
# Checking duplicate records

print("Remaining Duplicate Records:")

print(scaled_df.duplicated().sum())
```

```python
# Saving prepared dataset as CSV

scaled_df.to_csv(
    "Titanic_Prepared_Dataset.csv",
    index=False
)

print("Prepared Dataset Saved Successfully!")
print("File Name: Titanic_Prepared_Dataset.csv")
```
